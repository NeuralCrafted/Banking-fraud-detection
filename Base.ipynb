Checking whether each category actually tells you something about fraud, and whether the rare categories are safe to keep.

```python
overall = data['fraud_bool'].mean()
print("Overall fraud rate:", overall)
print()

for column in columns_to_check:
    result = data.groupby(column)['fraud_bool'].agg(['mean', 'sum', 'count'])
    print(result.sort_values('mean', ascending=False))
    print()
```
