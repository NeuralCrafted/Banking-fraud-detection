Check the negative values in the credit_risk_score(place holders or real)

```python
neg_values = data.loc[data['credit_risk_score'] < 0, 'credit_risk_score']
print(len(neg_values))#The total number of negative rows
print(neg_values.value_counts().head(10))#The most common negative values
print(neg_values.describe()) #min,max, spread of these rows
print(data.loc[data['credit_risk_score'] < 0, 'fraud_bool'].mean()) #fraud rate in these rows
```

The most common negative score, -1, appears only 488 times (3.4% of the negatives). The counts then fall gradually: 473, 440, 421, and so on. A placeholder would show a huge spike at one number, and there isn’t one.
The values spread across a wide range. They run from -170 to -1, with a median of -30. That looks like the low end of a continuous score, not a code.
The fraud rate is 0.39%, against 1.10% overall. At the overall rate you’d expect about 159 fraud cases among these rows, and you see about 56. So these rows are roughly a third as likely to be fraud, which is a real pattern.

So we replace -1 with NaN in these columns: prev_address_months_count, bank_months_count, current_address_months_count, session_length_in_minutes, device_distinct_emails_8w

```python
to_replace = ['prev_address_months_count', 'bank_months_count','current_address_months_count','session_length_in_minutes', 'device_distinct_emails_8w']
```

```python
for col in to_replace:
  data.loc[data[col] == -1, col] = np.nan
```

```python
for columns in to_replace:
  nulls_count = data[columns].isnull().sum()
  if nulls_count > 0:
    print(f"{columns} : {nulls_count}")
```

Lets split the data into the traning and testing Data.However we will use month to split the data and not a random split.This is because,fraud patterns and tactics change over time.For example in the data, it is clear that the fraud rate dips to 0.87% in month 2 and climbs to about 1.5% by month 7. A random split hides this drift, because both piles contain every month. A time split exposes it.

```python
train_data = data[data['month'] <= 5].copy()
test_data = data[data['month'] >= 6].copy()
```

```python
print(f"Training data: {len(train_data)}")
print(f"Testing data: {len(test_data)}")
```

Now encoding the non numerical columns

```python
cat_cols = ['payment_type', 'employment_status', 'housing_status', 'source', 'device_os']

for col in cat_cols:
    cats = sorted(train_data[col].unique())
    train_data[col] = pd.Categorical(train_data[col], categories=cats)
    test_data[col]  = pd.Categorical(test_data[col], categories=cats)
```

Baseline Models

```python
from sklearn.dummy import DummyClassifier
from sklearn.metrics import roc_auc_score, average_precision_score

dummy = DummyClassifier(strategy='prior')
dummy.fit(X_train, y_train)
p = dummy.predict_proba(X_test)[:, 1]

print("Dummy ROC-AUC:", roc_auc_score(y_test, p))
print("Dummy PR-AUC: ", average_precision_score(y_test, p))
```

```python
from sklearn.ensemble import HistGradientBoostingClassifier

hgb = HistGradientBoostingClassifier(
    categorical_features='from_dtype',
    class_weight= 'balanced',
    random_state= 10,
)
hgb.fit(X_train, y_train)
p = hgb.predict_proba(X_test)[:, 1]

print("HistGB ROC-AUC:", roc_auc_score(y_test, p))
print("HistGB PR-AUC: ", average_precision_score(y_test, p))
```

```python
sub_train = train_data[train_data['month'] <= 4]
val = train_data[train_data['month'] == 5]
```

```python
drop_cols = ['fraud_bool', 'month', 'customer_age', 'device_fraud_count']

X_sub = sub_train.drop(columns=drop_cols, errors='ignore')
y_sub = sub_train['fraud_bool']

X_val = val.drop(columns=drop_cols, errors='ignore')
y_val = val['fraud_bool']
```

```python
for cw in [None, 'balanced']:
    model = HistGradientBoostingClassifier(
        categorical_features='from_dtype',
        class_weight=cw,
        random_state=42,
    )
    model.fit(X_sub, y_sub)
    p = model.predict_proba(X_val)[:, 1]
    print(cw, "ROC-AUC:", roc_auc_score(y_val, p), "| PR-AUC:", average_precision_score(y_val, p))
```

Final Training

```python
drop_cols = ['fraud_bool', 'month', 'customer_age', 'device_fraud_count']

X_final_train = train_data.drop(columns=drop_cols, errors='ignore')
y_final_train = train_data['fraud_bool']
X_final_test  = test_data.drop(columns=drop_cols, errors='ignore')
y_final_test  = test_data['fraud_bool']

final_model = HistGradientBoostingClassifier(
    categorical_features='from_dtype',
    class_weight= 'balanced',
    random_state=42,
)
final_model.fit(X_final_train, y_final_train)
p = final_model.predict_proba(X_final_test)[:, 1]

print("ROC-AUC:", roc_auc_score(y_final_test, p))
print("PR-AUC: ", average_precision_score(y_final_test, p))
```
